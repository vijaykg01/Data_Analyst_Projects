```python
sns.set_context("notebook")
plt.figure(figsize=(10, 8))
sns.histplot(data=df, x ='Sales', bins = 100, kde = True, color = "skyblue")
plt.title('Distribution of Sales')
plt.xlabel('Sales')
plt.ylabel('Density')
plt.grid(True, alpha=0.2, linestyle='--')
plt.show()
```

```python
sns.set_context("notebook")
sns.barplot(data=df, y = 'Shipping.Cost', x = "Order.Priority", color = "blue")
plt.title('Distribution of Sales',fontsize = 15,  fontweight ="bold" )
plt.xlabel("Order_priority",fontsize = 12)
plt.ylabel("Shipping_cost", fontsize = 12)
plt.grid(True, alpha = 0.2, linestyle = '--')
plt.show()
```

```python
sns.set_context("notebook")
sns.countplot(data=df, x = "Segment", color = "orange")
plt.title('Number of orders by segment',fontsize = 14,  fontweight ="bold" )
plt.xlabel("Segments",fontsize = 12)
plt.ylabel("Count", fontsize = 12)
plt.grid(True, alpha = 0.2, linestyle = '--')
plt.show()
```

```python
sns.scatterplot(data = df, y = "Profit", x = "Sales", hue = "Segment", palette = "Set1")
plt.title('Sales VS Profit',fontsize = 14,  fontweight ="bold" )
plt.xlabel("Sales",fontsize = 12)
plt.ylabel("Profit", fontsize = 12)
plt.grid(True, alpha = 0.2, linestyle = '--')
plt.show()
```

```python
sns.set_context("paper")
sns.boxplot(data = df, x= "Ship.Mode", y= "Sales", hue = "Order.Priority")
plt.title("Sales acc. to ship_mode", fontweight = "bold")
```

```python
market = df["Region"].value_counts()
plt.pie(market.values, labels= market.index, autopct = "%1.1f%%")
```

```python
sns.set_context("paper")
sns.violinplot(data = df, x= "Order.Priority", y= "Sales", hue = 'Ship.Mode', inner= "box")
```

```python
num_cols = [ "Sales", "Profit", "Quantity", 'Row.ID', 'Shipping.Cost', 'weeknum']
sns.set_context("paper")
sns.heatmap(data = df[num_cols].corr(), annot = True, linewidth = 2)
```
