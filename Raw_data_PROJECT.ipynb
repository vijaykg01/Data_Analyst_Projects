```python
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
df = pd.read_excel(r"C:\Users\VIJAY K G\Downloads\Seaborn_Analysis_Dataset_750_Rows.xlsx")
df.head()
```

```python
# SCATTER PLOT
sns.set_style("whitegrid")
sns.set_context("paper")
sns.scatterplot(data = df,
                x = "Month",
                y = "Sales",
                hue = "Region",
                palette = "Set1",
                alpha = 0.8)
sns.despine()
```

```python
# STRIP PLOT
sns.set_style("whitegrid")
sns.set_context("paper")
sns.stripplot(data = df,
                x = "Month",
                y = "Sales",
                hue = "Region",
                palette = "Set2",
                dodge = True,
                jitter = True)        
```

```python
raw = pd.read_csv(r"C:\Users\VIJAY K G\Downloads\employees_500_rows.csv")
raw.head()
```

```python
# SWARM PLOT 
sns.swarmplot(data = raw,
              x = "designation",
              y = "salary",
              hue = "gender")
```

```python
# HISTOGRAM PLOT 
sns.histplot(data = raw,
             x = "salary",
             hue = "gender",
             multiple = "stack")
```

```python
# REGREESION PLOT 
raw["department_code"] = raw["department"].astype("category").cat.codes

sns.regplot(
    data=raw,
    x="department_code",
    y="salary",
    color="orange")
```

```python
# LINE PLOT
sns.set_context("talk")
sns.lineplot(data = raw,
              x = "department",
              y = "salary",
              hue = "gender")
```

```python
# JOINT PLOT
sns.jointplot(data = df,
              x = "Profit",
              y = "Sales",
              hue = "Region")
```

```python
# JOINT PLOT
sns.jointplot(data = df,
              x = "Profit",
              y = "Sales",
              kind = "reg")
```

```python
# BAR PLOT
import numpy as np

sns.set_context("paper")
sns.barplot(data = raw,
            x = "designation",
            y = "salary",
            hue = "gender",
            palette = ["blue", "orange"],
            estimator = np.median)
```

```python
# COUNT PLOT
sns.set_context("notebook")
sns.countplot(data = raw,
              x = "age",
              hue = "gender")
```

```python
# BOX PLOT
sns.boxplot(data = raw,
            x = "department",
            y = "salary",
            color = "brown")
```

```python
# VIOLIN PLOT
sns.violinplot(data = raw,
               x = "designation",
               y = "salary",
               hue = "gender",
               split = True,
               inner = "box")
```

```python
# VIOLIN PLOT
sns.violinplot(
        data=raw,
        x="gender",
        y="salary",
        palette="pastel",
        hue="gender",
        legend=False)

# SHOW INDIVIDUAL DATA POINTS
sns.swarmplot(
        data=raw,
        x="gender",
        y="salary",
        size=5,
        color="black")
```

```python
# KDE PLOT
sns.kdeplot(data = raw,
            x = "salary",
            fill = True)
```

```python
# RUG PLOT
sns.rugplot(
        data = raw,
        y = "salary",
        hue = "gender",
        height = 0.3)
```

```python
# HEAT MAP
numeric_values = ["Sales", "Quantity", "Discount", "Profit", "Rating", "Age"]
sns.heatmap(data = df[numeric_values].corr())
```

```python
# HEAT MAP
numeric_values = ["Sales", "Quantity", "Discount", "Profit", "Rating", "Age"]
sns.heatmap(data = df[numeric_values].corr(),
           annot = True,
           cmap = "Blues",
           linewidth = 2,
           linecolor = "black")
```

```python
# PAIR PLOT
sns.pairplot(data = df,
            diag_kind = "kde")
```

```python
# PAIR GRID
graph = sns.PairGrid(data = df)
graph.map_upper(sns.scatterplot)
graph.map_lower(sns.kdeplot)
graph.map_diag(sns.histplot)
```
