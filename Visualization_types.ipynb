```python
import matplotlib.pyplot as plt
```

```python
# LINE CHART

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, color = 'orange', label= 'sales data')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('line chart')
plt.legend()
plt.show()
```

```python
# BAR CHART(vertical)

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.bar(x, y, color = 'green', label= 'sales data')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Bar chart')
plt.legend()
plt.show()
```

```python
# BAR CHART(horizontal)

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.barh(x, y, color = 'green',label= 'sales data')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Bar chart')
plt.legend()
plt.show()
```

```python
# PIE CHART

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.pie(x, labels= y, autopct='%1.1f%%',colors=['yellow','orange','red','green'])
plt.title('pie chart')
plt.show()
```

```python
# HISTOGRAM

scores = [45,65,67,53,76,85,49,82,91,96,57]

plt.hist(scores, bins=5, color='purple', edgecolor='black')
plt.show()
```

```python
# SCATTER PLOT

hrs_studied = [1, 2, 3, 4, 5]
exam_scores = [60,65,70,75,80]
plt.scatter(hrs_studied, exam_scores, color= 'green', marker='o', label= 'student data')
plt.legend()
plt.grid(True)
plt.show()
```
